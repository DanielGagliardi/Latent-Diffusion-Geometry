# The ridge for a sine wave

Let
$$
p_\sigma(x)
=
(p * \varphi_\sigma)(x)
=
\int_{\mathbb{R}^d}
p(y)\,
\frac{1}{(2\pi \sigma^2)^{d/2}}
\exp\!\left(
-\frac{\|x-y\|^2}{2\sigma^2}
\right)\,dy.
$$

Then the score of the Gaussian-smoothed density is

$$
\nabla \log p_\sigma(x)
=
\frac{
\displaystyle
\int_{\mathbb{R}^d}
\frac{y-x}{\sigma^2}\,
p(y)\,
\varphi_\sigma(x-y)\,dy
}{
\displaystyle
\int_{\mathbb{R}^d}
p(y)\,
\varphi_\sigma(x-y)\,dy
}.
$$

Equivalently, if
$$
Y \sim p,
\qquad
X = Y + \sigma Z,
\qquad
Z \sim \mathcal N(0,I),
$$
then

$$
\nabla \log p_\sigma(x)
=
\frac{1}{\sigma^2}
\left(
\mathbb E[Y \mid X=x] - x
\right).
$$

## 1. System Setup and Definitions

Let $y \in \mathbb{R}^d$ be a random variable distributed according to an arbitrary data distribution $p(y)$. Let $x \in \mathbb{R}^d$ be the corresponding noise-perturbed variable generated via the conditional distribution:

$$p_\sigma(x|y) = \mathcal{N}(x; y, \sigma^2 I_d) = \frac{1}{(2\pi\sigma^2)^{d/2}} \exp\left( -\frac{\|x - y\|^2}{2\sigma^2} \right)$$

The marginal distribution $p_\sigma(x)$ is defined by the convolution:

$$p_\sigma(x) = \int_{\mathbb{R}^d} p(y) p_\sigma(x|y) \, dy$$

By Bayes' theorem, the posterior distribution of the clean data $y$ given the observed noisy vector $x$ is:

$$p_\sigma(y|x) = \frac{p(y) p_\sigma(x|y)}{p_\sigma(x)}$$


## 2. Main Theorem

The Hessian matrix of the log-density $\log p_\sigma(x)$ with respect to $x$ is given exactly by the expression:

$$\nabla_x^2 \log p_\sigma(x) = \frac{1}{\sigma^4} \operatorname{Cov}_{y \sim p_\sigma(y|x)}(y) - \frac{1}{\sigma^2} I_d$$

where $\operatorname{Cov}_{y \sim p_\sigma(y|x)}(y) \in \mathbb{R}^{d \times d}$ is the conditional covariance matrix of the clean data given the noisy observation:

$$\operatorname{Cov}_{y \sim p_\sigma(y|x)}(y) = \mathbb{E}_{y \sim p_\sigma(y|x)}\left[ y y^T \right] - \mathbb{E}_{y \sim p_\sigma(y|x)}[y] \mathbb{E}_{y \sim p_\sigma(y|x)}[y]^T$$


## 3.  Derivation

### Step 3.1: First-Order Derivative (The Score Function)
We begin by evaluating the spatial gradient of the log-marginal density:

$$\text{score}(x) = \nabla_x \log p_\sigma(x) = \frac{\nabla_x p_\sigma(x)}{p_\sigma(x)}$$

Differentiating the convolution integral under the integral sign yields:

$$\nabla_x p_\sigma(x) = \int_{\mathbb{R}^d} p(y) \nabla_x p_\sigma(x|y) \, dy$$

The gradient of the Gaussian kernel with respect to $x$ is:

$$\nabla_x p_\sigma(x|y) = p_\sigma(x|y) \left( \frac{y - x}{\sigma^2} \right)$$

Substituting this back into the gradient equation and normalizing by $p_\sigma(x)$ provides:

$$\begin{aligned}
\nabla_x \log p_\sigma(x) &= \int_{\mathbb{R}^d} \frac{p(y) p_\sigma(x|y)}{p_\sigma(x)} \left( \frac{y - x}{\sigma^2} \right) \, dy \\
&= \int_{\mathbb{R}^d} p_\sigma(y|x) \left( \frac{y - x}{\sigma^2} \right) \, dy \\
&= \frac{\mathbb{E}_{y \sim p_\sigma(y|x)}[y] - x}{\sigma^2}
\end{aligned}$$

> **Tweedie's Formula:**
> $$\nabla_x \log p_\sigma(x) = \frac{\mathbb{E}_{y \sim p_\sigma(y|x)}[y] - x}{\sigma^2}$$

### Step 3.2: Jacobian of the Posterior Expectation
To compute the Hessian, we must differentiate the score function with respect to $x$:

$$\nabla_x^2 \log p_\sigma(x) = \nabla_x \left( \frac{\mathbb{E}_{y \sim p_\sigma(y|x)}[y] - x}{\sigma^2} \right) = \frac{1}{\sigma^2} \nabla_x \mathbb{E}_{y \sim p_\sigma(y|x)}[y] - \frac{1}{\sigma^2} I_d$$

We isolate and evaluate the Jacobian matrix of the conditional expectation $\mu(x) = \mathbb{E}_{y \sim p_\sigma(y|x)}[y] = \int_{\mathbb{R}^d} y \, p_\sigma(y|x) \, dy$. Utilizing the identity $\nabla_x p_\sigma(y|x) = p_\sigma(y|x) \nabla_x \log p_\sigma(y|x)$, we obtain:

$$\nabla_x \mathbb{E}_{y \sim p_\sigma(y|x)}[y] = \int_{\mathbb{R}^d} y \left[ \nabla_x \log p_\sigma(y|x) \right]^T p_\sigma(y|x) \, dy$$

Expanding the gradient of the log-posterior:

$$\log p_\sigma(y|x) = \log p(y) + \log p_\sigma(x|y) - \log p_\sigma(x)$$

$$\nabla_x \log p_\sigma(y|x) = \frac{y - x}{\sigma^2} - \nabla_x \log p_\sigma(x)$$

Substituting this vector into the integral expression:

$$\begin{aligned}
\nabla_x \mathbb{E}_{y \sim p_\sigma(y|x)}[y] &= \int_{\mathbb{R}^d} y \left( \frac{y - x}{\sigma^2} - \nabla_x \log p_\sigma(x) \right)^T p_\sigma(y|x) \, dy \\
&= \frac{1}{\sigma^2} \mathbb{E}_{y \sim p_\sigma(y|x)}\left[y(y-x)^T\right] - \mathbb{E}_{y \sim p_\sigma(y|x)}[y] \left(\nabla_x \log p_\sigma(x)\right)^T
\end{aligned}$$

### Step 3.3: Second-Order Matrix Expansion
We now substitute Tweedie's formula for the remaining $\nabla_x \log p_\sigma(x)$ term:

$$\begin{aligned}
\nabla_x \mathbb{E}_{y \sim p_\sigma(y|x)}[y] &= \frac{1}{\sigma^2} \left( \mathbb{E}[y y^T] - \mathbb{E}[y]x^T \right) - \mathbb{E}[y] \left( \frac{\mathbb{E}[y]^T - x^T}{\sigma^2} \right) \\
&= \frac{1}{\sigma^2} \mathbb{E}[y y^T] - \frac{1}{\sigma^2} \mathbb{E}[y]x^T - \frac{1}{\sigma^2}\mathbb{E}[y]\mathbb{E}[y]^T + \frac{1}{\sigma^2}\mathbb{E}[y]x^T \\
&= \frac{1}{\sigma^2} \left( \mathbb{E}[y y^T] - \mathbb{E}[y]\mathbb{E}[y]^T \right) \\
&= \frac{1}{\sigma^2} \operatorname{Cov}_{y \sim p_\sigma(y|x)}(y)
\end{aligned}$$

### Step 3.4: Final Substitution
Inserting the isolated Jacobian of the expectation back into the primary Hessian equation yields the final expression:

$$\nabla_x^2 \log p_\sigma(x) = \frac{1}{\sigma^2} \left( \frac{1}{\sigma^2} \operatorname{Cov}_{y \sim p_\sigma(y|x)}(y) \right) - \frac{1}{\sigma^2} I_d = \frac{1}{\sigma^4} \operatorname{Cov}_{y \sim p_\sigma(y|x)}(y) - \frac{1}{\sigma^2} I_d$$

Definition (d-dimensional ridge).
Let $p: \mathbb{R}^D \rightarrow \mathbb{R}$ be twice continuously differentiable, with gradient $g(x)=\nabla \log{p(x)}$ and Hessian $H(x)$. Let

$$
H(x)=U(x) \Lambda(x) U(x)^{\top},
$$

where the eigenvalues satisfy

$$
\lambda_1(x) \geq \cdots \geq \lambda_d(x)>\lambda_{d+1}(x) \geq \cdots \geq \lambda_D(x),
$$

and let $V(x) \in \mathbb{R}^{D \times(D-d)}$ contain the eigenvectors associated with the smallest $D-d$ eigenvalues. Define the orthogonal projector onto the local normal space,

$$
L(x)=V(x) V(x)^{\top},
$$

and the projected gradient,

$$
G(x)=L(x) \nabla \log{p(x)} .
$$


The $d$-dimensional ridge of $p$ is

$$
R=\left\{x \in \mathbb{R}^D: G(x)=0, \lambda_{d+1}(x)<0\right\} .
$$


Equivalently, a point lies on the ridge if the density is stationary in the $D-d$ normal directions and has strictly negative curvature along those directions.

## Ridge detection in 2D using the log-density

We work with the **log-density**

$$
f(x,y) = \log p(x,y),
$$

rather than the density itself. This is preferable in generative models because $f$ is typically what is available (up to an additive constant), and the ridges of $p$ and $\log p$ coincide since

$$
\nabla \log p = \frac{\nabla p}{p},
\qquad
\nabla^2 \log p
=
\frac{\nabla^2 p}{p}
-
\frac{\nabla p\,\nabla p^\top}{p^2}.
$$

Throughout, let

$$
g =
\nabla f
=
\begin{pmatrix}
f_x\\
f_y
\end{pmatrix},
$$

and

$$
H =
\nabla^2 f
=
\begin{pmatrix}
f_{xx} & f_{xy}\\
f_{xy} & f_{yy}
\end{pmatrix}.
$$

---

## Step 1: Compute the Hessian eigenvalues

The eigenvalues of the Hessian are

$$
\lambda_{1,2}
=
\frac{f_{xx}+f_{yy}}{2}
\pm
\frac12
\sqrt{
(f_{xx}-f_{yy})^2
+
4f_{xy}^2
},
$$

where

$$
\lambda_1 \ge \lambda_2.
$$

The smaller eigenvalue $\lambda_2$ corresponds to the direction of greatest negative curvature (the local normal direction).

---

## Step 2: Compute the normal eigenvector

A convenient expression for the eigenvector associated with $\lambda_2$ is

$$
v_2
=
\frac{
\begin{pmatrix}
f_{xy}\\
\lambda_2-f_{xx}
\end{pmatrix}
}{
\sqrt{
f_{xy}^2
+
(\lambda_2-f_{xx})^2
}
}.
$$

Any unit vector differing by an overall sign is equally valid.

---

## Step 3: Ridge conditions

A point lies on a one-dimensional ridge if both of the following hold:

1. **Negative normal curvature**

$$
\lambda_2 < 0.
$$

2. **No gradient in the normal direction**

$$
g^\top v_2 = 0.
$$

Equivalently,

$$
f_x f_{xy}
+
f_y(\lambda_2-f_{xx})
=
0,
$$

which avoids explicitly normalizing the eigenvector.

In practice this condition is checked numerically:

$$
|g^\top v_2| < \varepsilon,
$$

for some small tolerance $\varepsilon$.

---

## Implementation recipe

Given a function returning the log-density $f(x,y)$:

1. Compute the gradient
   - $f_x$
   - $f_y$

2. Compute the Hessian
   - $f_{xx}$
   - $f_{xy}$
   - $f_{yy}$

3. Compute

$$
\Delta
=
\sqrt{
(f_{xx}-f_{yy})^2
+
4f_{xy}^2
}.
$$

4. Compute

$$
\lambda_2
=
\frac{f_{xx}+f_{yy}-\Delta}{2}.
$$

5. Reject the point if

$$
\lambda_2 \ge 0.
$$

6. Compute the (unnormalized) normal direction

$$
n=
\begin{pmatrix}
f_{xy}\\
\lambda_2-f_{xx}
\end{pmatrix}.
$$

7. Accept the point as lying on the ridge if

$$
\frac{|g^\top n|}{\|n\|}
<
\varepsilon.
$$

This formulation avoids a general eigendecomposition: only the closed-form eigenvalue formula and a single eigenvector expression are required.

```python
import numpy as np
import matplotlib.pyplot as plt
from scipy.spatial import cKDTree
from ipywidgets import interact, FloatSlider, ToggleButtons

X_DOMAIN = 4.0
Y_DOMAIN = 2.0
GRID_RES_X = 240
GRID_RES_Y = 120
N_CURVE_PTS = 1500

xs = np.linspace(-X_DOMAIN, X_DOMAIN, GRID_RES_X)
ys = np.linspace(-Y_DOMAIN, Y_DOMAIN, GRID_RES_Y)
Xg, Yg = np.meshgrid(xs, ys)
G = np.stack([Xg.ravel(), Yg.ravel()], axis=-1)  # (M,2) query grid

x_line = np.linspace(-X_DOMAIN, X_DOMAIN, 1000)


def sample_curve(amplitude, wavelength, measure="arc_length", n=N_CURVE_PTS, t_pad=2.0):
    """Samples of y = amplitude*sin(2*pi/wavelength * x) on a padded x-range.
    measure="arc_length": p uniform wrt arc length (equal-length pieces of curve equally likely).
    measure="x": p uniform wrt x (equal-width x-intervals equally likely, regardless of slope)."""
    k = 2 * np.pi / wavelength
    if measure == "arc_length":
        t_dense = np.linspace(-X_DOMAIN - t_pad, X_DOMAIN + t_pad, 20_000)
        speed = np.sqrt(1.0 + (amplitude * k * np.cos(k * t_dense)) ** 2)
        s = np.concatenate([[0.0], np.cumsum(0.5 * (speed[1:] + speed[:-1]) * np.diff(t_dense))])
        s_uniform = np.linspace(s[0], s[-1], n)
        t_uniform = np.interp(s_uniform, s, t_dense)
    else:  # measure == "x": samples spaced evenly in x itself
        t_uniform = np.linspace(-X_DOMAIN - t_pad, X_DOMAIN + t_pad, n)
    t_dense = np.linspace(-X_DOMAIN - t_pad, X_DOMAIN + t_pad, 20_000)  # for curve-distance KDTree
    pts = np.stack([t_uniform, amplitude * np.sin(k * t_uniform)], axis=-1)
    return pts, t_dense, k


def mixture_stats(X, Yc, sigma, batch=4000):
    """X: (M,2) query points, Yc: (N,2) curve quadrature points (equal weight).
    Returns posterior mean mu (M,2), posterior covariance cov (M,2,2), log p_sigma(x) (M,)."""
    N = Yc.shape[0]
    inv2s2 = 1.0 / (2.0 * sigma ** 2)
    log_norm_const = -np.log(N) - np.log(2 * np.pi * sigma ** 2)
    M = X.shape[0]
    mu = np.empty((M, 2))
    cov = np.empty((M, 2, 2))
    logp = np.empty(M)
    for start in range(0, M, batch):
        Xb = X[start:start + batch]
        diffs = Xb[:, None, :] - Yc[None, :, :]
        sqdist = np.sum(diffs ** 2, axis=-1)
        logk = -sqdist * inv2s2
        m = logk.max(axis=1, keepdims=True)
        w = np.exp(logk - m)
        wsum = w.sum(axis=1, keepdims=True)
        w /= wsum
        mu_b = w @ Yc
        c = Yc[None, :, :] - mu_b[:, None, :]
        cov_b = np.einsum('bn,bni,bnj->bij', w, c, c)
        mu[start:start + batch] = mu_b
        cov[start:start + batch] = cov_b
        logp[start:start + batch] = (m[:, 0] + np.log(wsum[:, 0])) + log_norm_const
    return mu, cov, logp


def compute_ridge_points(amplitude, sigma, wavelength, epsilon, measure):
    """Explicit epsilon-threshold ridge test (markdown Step 3/7):
    accept x if lambda2(x) < 0 and |g(x)^T v2(x)| < epsilon."""
    Yc, t_dense, k = sample_curve(amplitude, wavelength, measure=measure)
    mu, cov, logp = mixture_stats(G, Yc, sigma)
    score = (mu - G) / sigma ** 2               # g(x), Tweedie
    H = cov / sigma ** 4 - np.eye(2) / sigma ** 2  # Hessian, Step 3.4

    eigvals, eigvecs = np.linalg.eigh(H)  # ascending: eigvals[:,0] <= eigvals[:,1]
    lambda2 = eigvals[:, 0]
    v2 = eigvecs[:, :, 0]                 # unit eigenvector -> already normalized (n/||n||)

    g_dot_v2 = np.sum(score * v2, axis=1)  # g^T v2, since ||v2||=1 this equals g^T n / ||n||

    tree = cKDTree(np.stack([t_dense, amplitude * np.sin(k * t_dense)], axis=-1))
    dist_to_curve, _ = tree.query(G)

    # explicit epsilon criterion: |g^T v2| < epsilon, plus lambda2 < 0
    is_ridge = (lambda2 < 0) & (np.abs(g_dot_v2) < epsilon) & (dist_to_curve <= 1.0 * sigma)

    density_field = np.exp(logp).reshape(Xg.shape)
    return G[is_ridge], density_field, k


def plot_ridge(amplitude=1.0, sigma=0.3, wavelength=2 * np.pi, epsilon=0.05, measure="arc_length"):
    ridge_pts, density_field, k = compute_ridge_points(amplitude, sigma, wavelength, epsilon, measure)

    fig, ax = plt.subplots(figsize=(9, 5))
    im = ax.pcolormesh(Xg, Yg, density_field, shading="auto", cmap="viridis", alpha=0.3)
    fig.colorbar(im, ax=ax, fraction=0.046, pad=0.04, label=r"$p_\sigma(x)$")

    ax.plot(x_line, amplitude * np.sin(k * x_line), color="tab:blue", lw=1.5, label="true sine curve")
    if ridge_pts.shape[0]:
        ax.scatter(ridge_pts[:, 0], ridge_pts[:, 1], s=4, color="tab:red",
                   label=rf"ridge pts ($\lambda_2<0$, $|g^\top v_2|<\varepsilon$)")

    ax.set_xlim(-X_DOMAIN, X_DOMAIN)
    ax.set_ylim(-Y_DOMAIN, Y_DOMAIN)
    ax.set_aspect("equal")
    ax.set_xlabel("x")
    ax.set_ylabel("y")
    measure_label = "uniform wrt arc length" if measure == "arc_length" else "uniform wrt x"
    ax.set_title(f"Ridge vs. true curve ({measure_label}), $A={amplitude:.2f}$, $\\sigma={sigma:.3f}$, $\\lambda={wavelength:.3f}$, $\\varepsilon={epsilon:.3f}$")
    ax.legend(loc="upper right", fontsize=8)
    plt.tight_layout()
    plt.show()


interact(
    plot_ridge,
    amplitude=FloatSlider(value=1.0, min=0.1, max=2.0, step=0.05, description="amplitude", continuous_update=False),
    sigma=FloatSlider(value=0.3, min=0.05, max=1.0, step=0.01, description="sigma", continuous_update=False),
    wavelength=FloatSlider(value=2 * np.pi, min=1.0, max=16.0, step=0.5, description="wavelength", continuous_update=False),
    epsilon=FloatSlider(value=0.05, min=0.001, max=0.3, step=0.001, description="epsilon", continuous_update=False),
    measure=ToggleButtons(options=[("arc length", "arc_length"), ("x", "x")], description="p uniform wrt:"),
)
```
