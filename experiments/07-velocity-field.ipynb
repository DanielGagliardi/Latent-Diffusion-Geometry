# Velocity field for a circle

**Formulation of the Problem**

Let the ambient space be $\mathbb{R}^{d+1}$. The base distribution is the standard multivariate Gaussian:


$$p_0(x_0) = \mathcal{N}(x_0; 0, I_{d+1})$$

The target distribution $p_1(x_1)$ is a noisy distribution over the $d$-sphere $\mathbb{S}^d \subset \mathbb{R}^{d+1}$. A sample $x_1 \sim p_1$ is generated by sampling a point $u$ uniformly from $\mathbb{S}^d$ and adding Gaussian noise with variance $\sigma^2/d$ per dimension:


$$x_1 | u \sim \mathcal{N}\left(u, \frac{\sigma^2}{d} I_{d+1}\right), \quad u \sim U(\mathbb{S}^d)$$

Under the independent coupling, the linear flow path is defined as:


$$x_t = (1-t)x_0 + t x_1, \quad t \in [0,1]$$

The objective is to compute the exact marginal vector field $v_t(x)$, which is given by the conditional expectation:


$$v_t(x) = \mathbb{E}[x_1 - x_0 \mid x_t = x]$$

---

**Derivation of the Vector Field**

Let $\tilde{\sigma}^2 = \sigma^2/d$. Conditioned on the latent point $u \in \mathbb{S}^d$, the marginal distribution of $x_t$ is a linear combination of independent Gaussians:


$$x_t \mid u = (1-t)x_0 + t(u + \epsilon)$$


where $x_0 \sim \mathcal{N}(0, I_{d+1})$ and $\epsilon \sim \mathcal{N}(0, \tilde{\sigma}^2 I_{d+1})$.

Thus, the distribution of $x_t$ given $u$ is Gaussian:


$$x_t \mid u \sim \mathcal{N}(t u, \sigma_t^2 I_{d+1})$$


where the time-dependent variance is $\sigma_t^2 = (1-t)^2 + t^2 \tilde{\sigma}^2$.

We first compute the conditional expectation of the velocity $x_1 - x_0$ given both $x_t$ and $u$. Since $x_1 - x_0$, $x_t$, and $u$ are jointly Gaussian conditioned on $u$, we evaluate the covariance:


$$\text{Cov}(x_1 - x_0, x_t \mid u) = \text{Cov}(u + \epsilon - x_0, t(u + \epsilon) + (1-t)x_0 \mid u)$$

$$= t \text{Var}(\epsilon) - (1-t) \text{Var}(x_0) = t \tilde{\sigma}^2 - (1-t)$$

Using the properties of conditional Gaussians:


$$\mathbb{E}[x_1 - x_0 \mid x_t = x, u] = \mathbb{E}[x_1 - x_0 \mid u] + \frac{\text{Cov}(x_1 - x_0, x_t \mid u)}{\text{Var}(x_t \mid u)} (x - \mathbb{E}[x_t \mid u])$$

$$\mathbb{E}[x_1 - x_0 \mid x_t = x, u] = u + \frac{t \tilde{\sigma}^2 - (1-t)}{\sigma_t^2} (x - t u)$$

Regrouping terms, we obtain:


$$\mathbb{E}[x_1 - x_0 \mid x_t = x, u] = \frac{t \tilde{\sigma}^2 - (1-t)}{\sigma_t^2} x + \frac{1-t}{\sigma_t^2} u$$

To find $v_t(x)$, we marginalize over the latent variable $u$:


$$v_t(x) = \mathbb{E}_{u \mid x_t=x} \left[ \frac{t \tilde{\sigma}^2 - (1-t)}{\sigma_t^2} x + \frac{1-t}{\sigma_t^2} u \right] = \frac{t \tilde{\sigma}^2 - (1-t)}{\sigma_t^2} x + \frac{1-t}{\sigma_t^2} \mathbb{E}[u \mid x_t = x]$$

By Bayes' theorem, the posterior $p(u \mid x_t = x)$ is proportional to the likelihood $p(x_t = x \mid u)$:


$$p(u \mid x_t = x) \propto \exp\left( -\frac{\|x - tu\|^2}{2\sigma_t^2} \right) \propto \exp\left( \frac{t}{\sigma_t^2} \langle x, u \rangle \right)$$


Since $\|u\| = 1$, the posterior is exactly the von Mises-Fisher (vMF) distribution on $\mathbb{S}^d \subset \mathbb{R}^{d+1}$, with concentration parameter $\kappa_t(x) = \frac{t \|x\|}{\sigma_t^2}$ and mean direction $\mu = \frac{x}{\|x\|}$.

The expectation of a random vector distributed according to a vMF distribution in $\mathbb{R}^{d+1}$ is analytically given by:


$$\mathbb{E}[u \mid x_t = x] = \frac{I_{(d+1)/2}(\kappa_t(x))}{I_{(d-1)/2}(\kappa_t(x))} \frac{x}{\|x\|}$$


where $I_\nu$ denotes the modified Bessel function of the first kind of order $\nu$.

---

**Exact Velocity Field**

Substituting the expected value of $u$ back into the marginalization equation and replacing $\tilde{\sigma}^2$ with $\sigma^2/d$, the exact exact velocity field is:

$$v_t(x) = \left[ \frac{t \frac{\sigma^2}{d} - (1-t)}{(1-t)^2 + t^2 \frac{\sigma^2}{d}} + \frac{1-t}{(1-t)^2 + t^2 \frac{\sigma^2}{d}} \frac{1}{\|x\|} \frac{I_{(d+1)/2}\left( \frac{t \|x\|}{(1-t)^2 + t^2 \frac{\sigma^2}{d}} \right)}{I_{(d-1)/2}\left( \frac{t \|x\|}{(1-t)^2 + t^2 \frac{\sigma^2}{d}} \right)} \right] x$$
